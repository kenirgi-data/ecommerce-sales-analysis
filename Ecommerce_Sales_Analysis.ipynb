# E-Commerce Sales Analysis

## 📊 Overview
Analisis data penjualan e-commerce untuk memahami performa kategori produk.

```
import pandas as pd
import matplotlib.pyplot as plt
```

## 📁 Dataset

```
data = {
    'Category': ['Electronics', 'Fashion', 'Home', 'Electronics', 'Fashion', 'Home'],
    'Month': ['Jan', 'Jan', 'Jan', 'Feb', 'Feb', 'Feb'],
    'Sales': [200, 150, 100, 300, 180, 130]
}

df = pd.DataFrame(data)
df
```

## 📈 Analisis Penjualan per Kategori

```
category_sales = df.groupby('Category')['Sales'].sum()
category_sales
```

## 📊 Visualisasi

```
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.show()
```

## 💡 Insight
Kategori dengan penjualan tertinggi dapat menjadi fokus strategi bisnis.
